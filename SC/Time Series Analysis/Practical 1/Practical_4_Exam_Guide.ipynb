# Practical 4: Yearly Revenue Non-Seasonal Time Series Analysis
## Dataset: Yearly_revenue.xls (Annual Revenue Records)

---
### 📘 Beginner's Concept: Annual Data & Non-Seasonal Modeling
When time series data is recorded **annually** (once per year), there are no monthly or quarterly seasonal cycles. In statsmodels, we specify `period=1`, and we use **Holt's Linear Model** (Level + Trend) instead of seasonal Holt-Winters models.
---

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
df = pd.read_csv("Yearly_revenue.xls")
df.head(5)
```

```python
df.shape
```

```python
# Create sequential integer index for Year
df["Year"] = range(1, len(df) + 1)
df = df.set_index("Year")
df.head(5)
```

```python
sns.lineplot(data=df)
plt.title("Yearly Revenue Trend")
plt.show()
```

```python
# Non-seasonal decomposition (period=1)
result = seasonal_decompose(df, model="multiplicative", period=1)
result.plot()
plt.show()
```

```python
# Double Exponential Smoothing (Holt's Model for Trend)
from statsmodels.tsa.api import Holt
target_col = df.columns[0]
model_double = Holt(df[target_col])
model_double_fit = model_double.fit()
print(model_double_fit.params)
```

```python
# Stationarity Tests & First Differencing
from statsmodels.tsa.stattools import adfuller, kpss
diff_rev = df[target_col].diff().dropna()
print("First Diff ADF p-value:", adfuller(diff_rev)[1])
print("First Diff KPSS p-value:", kpss(diff_rev)[1])
```
